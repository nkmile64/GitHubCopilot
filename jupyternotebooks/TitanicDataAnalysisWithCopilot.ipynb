# Exploring Titanic data with Copilot
The titanic is a famous ship that sank in 1912 after hitting an iceberg. The data set contains information about the passengers on board, including their age, gender,class, and whether they survived or not. In this notebook, we will use Copilot to explore the data and find insights about the passengers.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)
```

```python
# Let's have an overview of the data by printing the first 5 rows
titanic.head()
```

q: What is the meaning of SibSp?
A: SibSp stands for "Siblings/Spouses Aboard". It indicates the number of siblings or spouses that a passenger had on board the Titanic. This information can be useful in analyzing the survival rates of passengers based on their family connections.

q: What is the meaning of Parch?
A: Parch stands for "Parents/Children Aboard". It indicates the number of parents or children that a passenger had on board the Titanic. Similar to SibSp, this information can help analyze survival rates based on family connections.

```python
# Let's compute how many passengers have survived and print the value
survived_count = titanic[titanic["Survived"] == 1]
print(f"Number of passengers who survived:\t{len(survived_count)}")

# Let's compute how many passengers have died and print the value
died_count = titanic[titanic["Survived"] == 0]
print(f"Number of passengers who died:\t\t{len(died_count)}")
```

```python
# Let's look at the correlation between the column Survived and the column Pclass
# And visualize the number of people per each class survived or not with a catplot
import seaborn as sns

sns.catplot(x="Pclass", hue="Survived", kind="count", data=titanic)
```
